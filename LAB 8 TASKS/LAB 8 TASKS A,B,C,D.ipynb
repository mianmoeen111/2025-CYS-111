```python
def greedy_coins(coins, amount):

    coins = sorted(coins, reverse=True)
    result = []

    for coin in coins:

        while amount >= coin:
            result.append(coin)
            amount -= coin

    return result


# Test cases

tests = [68, 41, 99]

for amount in tests:

    result = greedy_coins([25, 10, 5, 1], amount)

    print("Amount:", amount)
    print("Coins:", result)
    print("Count:", len(result))
    print()
```

```python
def select_activities(activities, rule):
    # Sort activities according to the selected rule
    if rule == "end":
        sorted_activities = sorted(activities, key=lambda x: x[2])

    elif rule == "start":
        sorted_activities = sorted(activities, key=lambda x: x[1])

    elif rule == "duration":
        sorted_activities = sorted(activities, key=lambda x: x[2] - x[1])

    selected = []
    last_end = 0

    # Select activities that do not overlap
    for activity in sorted_activities:
        name, start, end = activity

        if start >= last_end:
            selected.append(activity)
            last_end = end

    return selected


# Given activities
activities = [
    ("A", 1, 4),
    ("B", 3, 5),
    ("C", 0, 6),
    ("D", 5, 7),
    ("E", 3, 9),
    ("F", 5, 9),
    ("G", 6, 10),
    ("H", 8, 11),
    ("I", 8, 12),
    ("J", 2, 14),
    ("K", 12, 16)
]

# Rule 1: Earliest end time
result1 = select_activities(activities, "end")
print("Earliest End Time:")
print(result1)
print("Number of activities:", len(result1))

print()

# Rule 2: Earliest start time
result2 = select_activities(activities, "start")
print("Earliest Start Time:")
print(result2)
print("Number of activities:", len(result2))

print()

# Rule 3: Shortest duration
result3 = select_activities(activities, "duration")
print("Shortest Duration:")
print(result3)
print("Number of activities:", len(result3))
```

```python
# ==========================================
# TASK D: GREEDY AND A* PATHFINDING
# ==========================================

# Step 1: Create the map

grid = [
    "..........",
    "...#......",
    "...#......",
    "...#####..",
    ".S.#...#E.",
    "...#...#..",
    ".......##.",
    "..####.##.",
    ".......##.",
    ".........."
]

print("Original Map:")
for row in grid:
    print(row)


# Step 2: Find the starting and ending positions

def find(grid, letter):
    for r in range(len(grid)):
        for c in range(len(grid[r])):
            if grid[r][c] == letter:
                return (r, c)


# Step 3: Calculate Manhattan distance

def distance(cell, goal):
    return abs(cell[0] - goal[0]) + abs(cell[1] - goal[1])


# Step 4: Find neighboring cells

def neighbors(grid, cell):
    r, c = cell
    result = []

    # Up, Down, Left, Right
    directions = [
        (-1, 0),
        (1, 0),
        (0, -1),
        (0, 1)
    ]

    for dr, dc in directions:
        nr = r + dr
        nc = c + dc

        if (0 <= nr < len(grid) and
            0 <= nc < len(grid[0]) and
            grid[nr][nc] != "#"):

            result.append((nr, nc))

    return result


# Step 5: Implement Greedy Best-First Search

def greedy(grid):

    start = find(grid, "S")
    goal = find(grid, "E")

    frontier = [start]
    came_from = {start: None}
    visited = []

    while len(frontier) > 0:

        best = frontier[0]

        # Select the cell with the smallest heuristic
        for cell in frontier:
            if distance(cell, goal) < distance(best, goal):
                best = cell

        frontier.remove(best)
        visited.append(best)

        if best == goal:
            break

        for nxt in neighbors(grid, best):

            if nxt not in came_from:
                came_from[nxt] = best
                frontier.append(nxt)

    # Check whether the goal was reached
    if goal not in came_from:
        return visited, []

    # Build the path
    path = []
    cell = goal

    while cell is not None:
        path.append(cell)
        cell = came_from[cell]

    path.reverse()

    return visited, path


# Step 6: Implement A* Search

def a_star(grid):

    start = find(grid, "S")
    goal = find(grid, "E")

    frontier = [start]
    came_from = {start: None}
    cost = {start: 0}
    visited = []

    while len(frontier) > 0:

        best = frontier[0]

        # Select the cell with the smallest g + h
        for cell in frontier:

            if (cost[cell] + distance(cell, goal) <
                cost[best] + distance(best, goal)):

                best = cell

        frontier.remove(best)
        visited.append(best)

        if best == goal:
            break

        for nxt in neighbors(grid, best):

            new_cost = cost[best] + 1

            if nxt not in cost or new_cost < cost[nxt]:

                cost[nxt] = new_cost
                came_from[nxt] = best

                if nxt not in frontier:
                    frontier.append(nxt)

    # Check whether the goal was reached
    if goal not in came_from:
        return visited, []

    # Build the path
    path = []
    cell = goal

    while cell is not None:
        path.append(cell)
        cell = came_from[cell]

    path.reverse()

    return visited, path


# Step 7: Display the map with visited cells and path

def show(grid, visited, path):

    for r in range(len(grid)):

        line = ""

        for c in range(len(grid[r])):

            ch = grid[r][c]

            if ch == ".":

                if (r, c) in path:
                    ch = "*"

                elif (r, c) in visited:
                    ch = "o"

            line = line + ch

        print(line)


# Step 8: Run Greedy Search

print("\n========== GREEDY SEARCH ==========")

visited, path = greedy(grid)

show(grid, visited, path)

print("Visited cells:", len(visited))

if path:
    print("Path length:", len(path) - 1)
else:
    print("No path found")


# Step 9: Run A* Search

print("\n========== A* SEARCH ==========")

visited, path = a_star(grid)

show(grid, visited, path)

print("Visited cells:", len(visited))

if path:
    print("Path length:", len(path) - 1)
else:
    print("No path found")
```
